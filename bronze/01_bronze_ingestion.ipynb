```python
%sql
use catalog `e-commerce`;
use schema bronze;
```

```python
geolocation_df=spark.read.csv("s3://dhanya-e-commerce-datasets/olist_geolocation_dataset.csv",header=True)
orders_df=spark.read.csv("s3://dhanya-e-commerce-datasets/olist_orders_dataset.csv",header=True)
order_items_df=spark.read.csv("s3://dhanya-e-commerce-datasets/olist_order_items_dataset.csv",header=True)
customers_df=spark.read.csv("s3://dhanya-e-commerce-datasets/olist_customers_dataset.csv",header=True)
products_df=spark.read.csv("s3://dhanya-e-commerce-datasets/olist_products_dataset.csv",header=True)
product_category_df=spark.read.csv("s3://dhanya-e-commerce-datasets/product_category_name_translation.csv",header=True)
order_payments_df=spark.read.csv("s3://dhanya-e-commerce-datasets/olist_order_payments_dataset.csv",header=True)
sellers_df=spark.read.csv("s3://dhanya-e-commerce-datasets/olist_sellers_dataset.csv",header=True)
order_reviews_df = spark.read.format("csv").option("header", "true").option("multiLine", "true").option("quote", '"').option("escape", '"').load("s3://dhanya-e-commerce-datasets/olist_order_reviews_dataset.csv")

```

```python
customers_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.customers")
geolocation_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.geolocation")
order_items_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.order_items")
order_payments_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.order_payments")
order_reviews_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.order_reviews")
orders_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.orders")
product_category_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.product_category")
products_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.products")
sellers_df.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("`e-commerce`.bronze.sellers")
```
